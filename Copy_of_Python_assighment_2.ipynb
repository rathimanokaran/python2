```python
##list (creation,modification and access)
## create a list named age_list with five integer elements##

age_list=[24,25,26,27,28]
print(age_list)
```

```python
## b. create a list named name_list with five string elements##

name_list=["Rathi","srimathi","Mano","Eniyan","Yuva"]
print(name_list)
```

```python
A.## Append the string##
##Append the string "Yazhini"to name_list##

name_list.append("Yazhini")
print(name_list)
```

```python
##B. Insert the element##
##insert the element 30 at index 2 in age_list##

age_list=[24,25,26,27,28]
age_list.insert(2,30)
print(age_list)
```

```python
##C.Remove the string "Yazhini"From (name_list)##

name_list.remove("Yazhini")
print(name_list)
```

```python
##D.pop the last element from age_list##
age_list.pop()
print(age_list)
```

```python
##E.Extend age_list with additional ages [29,30,26]##

age_list.extend([29,30,26])
print(age_list)
```

```python
## F. sort age _list in descending order##
age_list.sort(reverse=True)
print(age_list)
```

```python
## G. Find Max age,Min age and Sum of all ages##

print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```
